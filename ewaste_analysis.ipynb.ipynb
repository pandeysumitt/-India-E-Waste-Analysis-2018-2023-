```python
# Clean all numeric columns at once
cols_to_clean = [
    'E-Waste Generated (Tonnes)',
    'E-Waste Collected (Tonnes)',
    'Collection Rate (%)',
    'Recycled (Tonnes)',
    'Recycling Efficiency (%)',
    'Hazardous Waste (Tonnes)'
]

for col in cols_to_clean:
    df[col] = df[col].astype(str).str.replace(',', '', regex=False).str.replace('%', '', regex=False).astype(float)
```

Analysis

```python
total_generated = df['E-Waste Generated (Tonnes)'].sum()
print(f"Total E-Waste Generated (2018-2023): {total_generated:,.2f} Tonnes")
```

```python
total_collected = df['E-Waste Collected (Tonnes)'].sum()
total_generated = df['E-Waste Generated (Tonnes)'].sum()
gap = total_generated - total_collected
```

```python
print(f"Total E-Waste Collected (2018-23) : {total_collected:,.2f} Tonnes" )
print(f"Total E-Waste Generated (2018-23): {total_generated:,.2f} Tonnes")
print(f"Gap : {gap:,.2f} Tonnes")
```

```python
# Average Collection Rate Nationally
avg_collection_rate = df['Collection Rate (%)'].mean()
print(f"Average Collection Rate Nationally: {avg_collection_rate:.2f}%")
```

```python
# State-wise:  Top 10 states by e-waste generated
top10_states = df.groupby('State')['E-Waste Generated (Tonnes)'].sum().sort_values(ascending=False).head(10)
print("Top 10 States by E-Waste Generated:")
print(top10_states)
```

```python
# Bottom 5 states by collection rate
bottom5_states = df.groupby('State')['Collection Rate (%)'].mean().sort_values().head(5)
print("\nBottom 5 States by Collection Rate:")
print(bottom5_states)
```

```python
# State with highest hazardous waste
top_hazardous_state = df.groupby('State')['Hazardous Waste (Tonnes)'].sum().idxmax()
top_hazardous_value = df.groupby('State')['Hazardous Waste (Tonnes)'].sum().max()
print(f"\nState with Highest Hazardous Waste: {top_hazardous_state}- {top_hazardous_value:,.2f} Tonnes")
```

```python
# Year-over-year e-waste generation trend
yoy = df.groupby('Year')['E-Waste Generated (Tonnes)'].sum()
print("Year-over-year E-Waste Generation :")
print(yoy)
```

```python
# Collection rate improvement per year
yearly_cr = df.groupby('Year')['Collection Rate (%)'].mean()
yearly_cr_diff = yearly_cr.diff()
print("\nCollection Rate Improvement Per Year:")
print(yearly_cr_diff)
```

```python
# Which year had the COVID dip
covid_dip_year = yearly_cr_diff.idxmin()
print(f"\nCovid Dip Year: {covid_dip_year} ({yearly_cr_diff.min():.2f}% drop)")
```

```python
# Category contributing most to total e-waste
top_category = df.groupby('Category')['E-Waste Generated (Tonnes)'].sum().sort_values(ascending=False)
print("Category Contributing Most to Total E-Waste:")
print(top_category)
```

```python
#  Most hazardous category
most_hazardous_cat = df.groupby('Category')['Hazardous Waste (Tonnes)'].sum().idxmax()
most_hazardous_value = df.groupby('Category')['Hazardous Waste (Tonnes)'].sum().max()
print(f"\nMost Hazardous Category: {most_hazardous_cat} -{most_hazardous_value:,.2f} Tonnes")
```

```python
# Best recycling efficiency by category
best_recycling = df.groupby('Category')['Recycling Efficiency (%)'].mean().sort_values(ascending=False)
print("\nBest Recycling Efficiency by Category:")
print(best_recycling)
```

```python
#  Total generated vs collected by region
region_summary = df.groupby('Region')[['E-Waste Generated (Tonnes)', 'E-Waste Collected (Tonnes)']].sum()
region_summary['Gap'] = region_summary['E-Waste Generated (Tonnes)'] - region_summary['E-Waste Collected (Tonnes)']
region_summary['Collection Rate (%)'] = (region_summary['E-Waste Collected (Tonnes)']/ region_summary['E-Waste Generated (Tonnes)']) *100
print(region_summary)
```

```python
# Recycling efficiency by region
region_recycling = df.groupby('Region')['Recycling Efficiency (%)'].mean().sort_values(ascending= False)
print("\nRecycling Efficiency by Region:")
print(region_recycling)
```

# Visualization

```python
# Top 10 States — Horizontal Bar Chart
top10_states = df.groupby('State')['E-Waste Generated (Tonnes)'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
colors = ['#C00000' if i==0 else '#2E75B6' for i in range(len(top10_states))]
top10_states.sort_values().plot(kind='barh', color = colors)
plt.title('Top 10 E-Waste Generating State', fontsize=14, fontweight='bold')
plt.xlabel('Total E-Waste Generated (Tonnes)')
plt.tight_layout()
plt.show()
```

```python
# Year Trend — Line Chart
yoy = df.groupby('Year')['E-Waste Generated (Tonnes)'].sum()

plt.figure(figsize=(10,5))
plt.plot(yoy.index, yoy.values, marker='o', color ='#2E75B6', linewidth= 2.5, markersize= 8)
plt.fill_between(yoy.index, yoy.values, alpha= 0.1, color = '#2E75B6')
plt.title('Year-over-Year E-Waste Generation Trend', fontsize=14, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Total E-Waste Generated (Tonnes)')
plt.xticks(yoy.index)
plt.tight_layout()
plt.show()
```

```python
#  Category Share — Pie Chart
cat_share = df.groupby('Category')['E-Waste Generated (Tonnes)'].sum()

plt.figure(figsize=(10,7))
plt.pie(cat_share, labels = cat_share.index, autopct ='%1.1f%%',
        startangle=140, colors=sns.color_palette('Set2',len(cat_share)))
plt.title('E-Waste Distribution by Category', fontsize =14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
#Collection Rate Trend — Line Chart
yearly_cr = df.groupby('Year')['Collection Rate (%)'].mean()

plt.figure(figsize=(10,5))
plt.plot(yearly_cr.index, yearly_cr.values, marker='s', color ='green',linewidth=2.5, markersize = 8)
for x, y in zip(yearly_cr.index, yearly_cr.values):
  plt.annotate(f'{y:.1f}%', (x,y), textcoords="offset points", xytext =(0,10), ha='center')
plt.title('E-Waste Collection Rate Trend', fontsize = 14, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Avg Collection Rate (%)')
plt.xticks(yearly_cr.index)
plt.tight_layout()
plt.show()
```

```python
#Region — Generated vs Collected Grouped Bar
region_summary = df.groupby('Region')[['E-Waste Generated (Tonnes)', 'E-Waste Collected (Tonnes)']].sum()

region_summary.plot(kind='bar', figsize=(12,6), color = ['#2E75B6', '#70AD47'],
                    width = 0.7)
plt.title('E-Waste Generated vs E-Waste Collected by Region', fontsize =14, fontweight ='bold')
plt.xlabel('Region')
plt.ylabel('Tonnes')
plt.xticks(rotation=45)
plt.legend(['Generated', 'Collected'])
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap — Seaborn
numeric_cols =df[['E-Waste Generated (Tonnes)', 'E-Waste Collected (Tonnes)',
                'Collection Rate (%)', 'Recycled (Tonnes)',
                'Recycling Efficiency (%)', 'Hazardous Waste (Tonnes)']]

plt.figure(figsize=(10,7))
sns.heatmap(numeric_cols.corr(), annot =True, fmt= '.2f',
            cmap= 'Blues', linewidths = 0.5,square= True)
plt.title('Correlation Matrix - E-Waste Variables', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
#Hazardous Waste by Region — Seaborn Bar
hazardous_region = df.groupby('Region')['Hazardous Waste (Tonnes)'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=hazardous_region.index, y=hazardous_region.values, palette='Reds_r')
plt.title('Hazardous E-Waste by Region', fontsize=14, fontweight='bold')
plt.xlabel('Region')
plt.ylabel('Hazardous Waste (Tonnes)')
plt.tight_layout()
plt.show()
```
